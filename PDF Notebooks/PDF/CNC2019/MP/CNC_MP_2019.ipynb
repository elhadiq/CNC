## Médian d’une liste de nombres
<div class="alert alert-info" role="alert"> 

- **(1 pt) Q 1**- Écrire la fonction grands(L,x) qui reçoit en paramètres une liste de nombres L, et un élément x de L. 
    - La fonction renvoie le nombre d’éléments de L qui sont supérieurs strictement à x.
    </div>

````python
#Réponse
def grands(L,x):
    n=0
    for  e in L:
        if e>x:
            n+=1
    return n
````

````python
#Test
L=[1,10,45,3,8]
print(grands(L,3))
````

- **0.5 pt Q 3**- Écrire la fonction **petits(L,x)** qui reçoit en paramètres une liste de nombres L, et un élément x de L. 
    - La fonction renvoie le nombre d’éléments de L qui sont inférieurs strictement à x
    <div>

````python
#Réponse
def petits(L,x):
    n=0
    for  e in L:
        if e<x:
            n+=1
    return n
````

**L** est une liste de taille **n** qui contient des nombres, et **m** un élément de **L**.

L’élément **m** est un médian de **L**, si les deux conditions suivantes sont vérifiées :

- Le nombre d’éléments de **L**, qui sont supérieurs strictement à m, est inférieur ou égale à n/2
- Le nombre d’éléments de **L**, qui sont inférieurs strictement à m, est inférieur ou égale à n/2


**Exemple** : On considère la liste L = [ 25 , 12 , 6 , 17, 3 , 10 , 20 , 12 , 15 , 38 ], de taille n=10.

L’élément 12 est un médian de L, car :
- 3 éléments de L sont supérieurs strictement à 12, et 3 ≤ n/2 ;
- 5 éléments de L sont inférieurs strictement à 12, et 5 ≤ n/2.

<div class="alert alert-info" role="alert"> 

- **Q 4**- Écrire la fonction **median(L)** qui reçoit en paramètre une liste de nombres L non vide, et qui renvoie un élément médian de la liste L.
<div>

````python
# Réponse
def median(L):
    n=len(L)
    assert n>0
    for m in L:
        if grands(L,m)<=n//2 and petits(L,m)<=n//2:
            return m
````

````python
#Test
median([10,3,5,9,11,2,6])
````

<div class="alert alert-success" role="alert">
    
**Réponse**
    
Soit $n\rightarrow T(n)$, la complexité de la fonction **median**, alors cette fonction prends en paramètre un entier n qui designe la taille de la liste passée en argument pour la focntion **medians**, et donc la complexité T vérifié:
    $$T(n)=C2+\Sigma_1^n (C3 + c(n) + c(n))=C2+C3 n +2n c(n)=O(n^2)$$
