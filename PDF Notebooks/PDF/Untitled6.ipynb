strictement entre 100 et 1000, triés dans l’ordre décroissant des nombres de villes.

```SQL
SELECT nom_pays, count(*) as nombre_de_villes
FROM Villes JOIN Pays
GROUP BY cod_pays
HAVING nombre_de_villes>100 and nombre_de_villes<1000
Order by nombre_de_villes desc
```

Q 2. b- Écrire la fonction list_voisines(i,R), qui reçoit en paramètres une ville i d’un réseau
routier représenté par la matrice symétrique R. La fonction renvoie la liste de toutes les villes voisines à la
ville i.

````python
def list_voisines(i,R):
    voisins=[]
    for j in range(R[i]):
        if R[i][j]==1:
            voisines.append(j)
    return voisines
````

````python
def degre(i,R):
    return len(list_voisines(i,R))
````

````python
def list_degres(R):
    L=[]
    for i in range(len(R)):
        L.append((i,degre(i,R)))
    return L
````

````python
def puissance(R,n):
    if n%2==0:
        C=puissance(R,n/2)
        return C.dot(C)
    else:
        C=puissance(R,(n-1)/2)
        return (C.dot(C)).dot(R)
````
